# VAJRA: Phase 3 - Exploratory Data Analysis (EDA)

In this notebook, we parse the raw data downloaded in Phase 2:
1. **ERA5 Atmospheric Data** (NetCDF)
2. **IMD Gridded Data** (Binary .grd converted to xarray)

```python
import os
import xarray as xr
import matplotlib.pyplot as plt
import imdlib as imd
import numpy as np

data_dir = '../data/raw'
era5_path = os.path.join(data_dir, 'era5_extracted/data_stream-oper_stepType-instant.nc')
imd_rain_path = os.path.join(data_dir, 'rain_2023.grd')
```

## 1. Parse and Visualize ERA5 Data
ERA5 provides critical atmospheric context (CAPE, winds, temperature).

```python
if os.path.exists(era5_path):
    # Load ERA5 dataset
    ds_era5 = xr.open_dataset(era5_path)
    print(ds_era5)
    
    # Plot Convective Available Potential Energy (CAPE) for a specific time
    plt.figure(figsize=(10, 6))
    # Plot the first time step
    ds_era5['cape'].isel(valid_time=0).plot(cmap='viridis')
    plt.title('ERA5 CAPE - July 1, 2023 (First Time Step)')
    plt.show()
else:
    print("ERA5 dataset not found. Please ensure the download script finished successfully.")
```

## 2. Parse and Visualize IMD Gridded Rainfall
We use `imdlib` to parse the `.grd` file.

```python
if os.path.exists(imd_rain_path):
    # Load IMD Rainfall data (Daily)
    data_rain = imd.open_data('rain', 2023, 2023, 'yearwise', data_dir)
    ds_rain = data_rain.get_xarray()
    print(ds_rain)
    
    # Plot rainfall for July 1, 2023 (Day 181 roughly)
    plt.figure(figsize=(10, 6))
    # Note: time index might need exact mapping, we'll plot the index 181
    ds_rain['rain'].isel(time=181).plot(cmap='Blues', vmin=0, vmax=50)
    plt.title('IMD Daily Rainfall - July 1, 2023')
    plt.show()
else:
    print("IMD Rainfall dataset not found.")
```
